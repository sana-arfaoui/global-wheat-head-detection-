# new import et setup

```python
import os
import random
import json
import ast
import numpy as np
import pandas as pd
import torch
import gc
from mmcv import Config
from mmdet.apis import set_random_seed, train_detector, single_gpu_test
from mmdet.models import build_detector
from mmdet.datasets import build_dataset, build_dataloader
from mmcv.runner import load_checkpoint

# Seed pour reproductibilité
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)

# Hyperparamètres
IMG_SIZE = 800
BATCH_SIZE = 8
EPOCHS = 50

# Dataset
IMG_DIR = "/kaggle/input/global-wheat-detection/train"
CSV_PATH = "/kaggle/input/global-wheat-detection/train.csv"

# Output
BASE_OUT = "/kaggle/working/GWHD_FasterRCNN_SwinT"
os.makedirs(BASE_OUT, exist_ok=True)

# Checkpoint Swin-T tiny
checkpoint_url = "https://github.com/SwinTransformer/storage/releases/download/v1.0.0/swin_tiny_patch4_window7_224.pth"
checkpoint_path = os.path.join(BASE_OUT, "swin_tiny_patch4_window7_224.pth")

if not os.path.exists(checkpoint_path):
    print("📥 Téléchargement du checkpoint Swin-T...")
    import urllib.request
    urllib.request.urlretrieve(checkpoint_url, checkpoint_path)
    print("✅ Téléchargement terminé")
else:
    print("✅ Checkpoint déjà téléchargé")
```

Conversion CSV → COCO


```python
def convert_to_coco(df, image_dir, output_json):
    images, annotations = [], []
    ann_id = 1

    for img_id, group in df.groupby("image_id"):
        file_name = f"{img_id}.jpg"
        img_path = os.path.join(image_dir, file_name)
        if not os.path.exists(img_path):
            continue

        width = int(group.iloc[0]["width"])
        height = int(group.iloc[0]["height"])

        images.append({
            "id": img_id,
            "file_name": file_name,
            "width": width,
            "height": height
        })

        for _, row in group.iterrows():
            bbox = ast.literal_eval(row["bbox"])
            annotations.append({
                "id": ann_id,
                "image_id": img_id,
                "category_id": 1,
                "bbox": bbox,
                "area": bbox[2] * bbox[3],
                "iscrowd": 0
            })
            ann_id += 1

    coco = {
        "images": images,
        "annotations": annotations,
        "categories": [{"id": 1, "name": "wheat"}]
    }

    with open(output_json, "w") as f:
        json.dump(coco, f)
```

# Préparer les folds LODO


```python
df = pd.read_csv(CSV_PATH)
countries = df["source"].unique()
fold_info = []

for fold_idx, test_country in enumerate(countries, 1):
    df_test = df[df.source == test_country]
    df_train_full = df[df.source != test_country]

    train_parts, val_parts = [], []

    for c in df_train_full.source.unique():
        df_c = df_train_full[df_train_full.source == c]
        ids = df_c.image_id.unique().tolist()
        random.shuffle(ids)
        n_val = int(0.1 * len(ids))
        val_ids = ids[:n_val]

        val_parts.append(df_c[df_c.image_id.isin(val_ids)])
        train_parts.append(df_c[~df_c.image_id.isin(val_ids)])

    df_train = pd.concat(train_parts)
    df_val = pd.concat(val_parts)

    fold_dir = os.path.join(BASE_OUT, f"fold_{fold_idx}_{test_country}")
    os.makedirs(fold_dir, exist_ok=True)

    convert_to_coco(df_train, IMG_DIR, f"{fold_dir}/train.json")
    convert_to_coco(df_val, IMG_DIR, f"{fold_dir}/val.json")
    convert_to_coco(df_test, IMG_DIR, f"{fold_dir}/test.json")

    fold_info.append({
        "fold_idx": fold_idx,
        "test_country": test_country,
        "train_domains": [c for c in countries if c != test_country],
        "n_train": df_train.image_id.nunique(),
        "n_val": df_val.image_id.nunique(),
        "n_test": df_test.image_id.nunique(),
        "fold_dir": fold_dir
    })

print("✅ Folds LODO prêts")
```

Charger config Faster R-CNN + Swin-T tiny


```python
cfg_base = Config.fromfile(
    "/kaggle/working/mmdetection/configs/faster_rcnn/faster_rcnn_r50_fpn_1x_coco.py"
)
print("✅ Config Faster R-CNN chargée")

# Remplacer backbone R50 par Swin-T tiny
cfg_base.model.backbone = dict(
    type='SwinTransformer',
    embed_dims=96,
    depths=[2, 2, 18, 2],
    num_heads=[3, 6, 12, 24],
    window_size=7,
    mlp_ratio=4.,
    qkv_bias=True,
    drop_rate=0.,
    attn_drop_rate=0.,
    drop_path_rate=0.2,
    patch_norm=True,
    out_indices=(0, 1, 2, 3),
    pretrain_style='official',
    init_cfg=dict(type='Pretrained', checkpoint=checkpoint_path)
)

# FPN doit correspondre aux sorties du Swin
cfg_base.model.neck.in_channels = [96, 192, 384, 768]
cfg_base.model.neck.out_channels = 256

# Nombre de classes pour bbox head
cfg_base.model.roi_head.bbox_head.num_classes = 1

```

 Préparer la config pour un fold

```python
def prepare_cfg(cfg, fold_dir):
    cfg = cfg.copy()
    cfg.work_dir = os.path.join(fold_dir, "work_dir")

    # Dataset
    for split in ["train", "val", "test"]:
        cfg.data[split].ann_file = f"{fold_dir}/{split}.json"
        cfg.data[split].img_prefix = IMG_DIR
        cfg.data[split].classes = ("wheat",)

    cfg.data.samples_per_gpu = BATCH_SIZE
    cfg.data.workers_per_gpu = 0

    # Optimizer
    cfg.optimizer = dict(
        type="AdamW",
        lr=5e-5,
        betas=(0.9, 0.999),
        weight_decay=0.05
    )

    cfg.lr_config = dict(
        policy="step",
        warmup="linear",
        warmup_iters=1000,
        warmup_ratio=0.001,
        step=[35, 45]
    )

    cfg.runner = dict(type="EpochBasedRunner", max_epochs=EPOCHS)

    cfg.evaluation = dict(
        interval=1,
        metric="bbox",
        save_best="bbox_mAP"
    )

    cfg.checkpoint_config = dict(interval=5, max_keep_ckpts=2)

    # Pipeline minimal
    cfg.data.train.pipeline = [
        dict(type="LoadImageFromFile"),
        dict(type="LoadAnnotations", with_bbox=True),
        dict(type="Resize", img_scale=(IMG_SIZE, IMG_SIZE), keep_ratio=False),
        dict(type="RandomFlip", flip_ratio=0.0),
        dict(type="Normalize", **cfg.img_norm_cfg),
        dict(type="Pad", size_divisor=32),
        dict(type="DefaultFormatBundle"),
        dict(type="Collect", keys=["img", "gt_bboxes", "gt_labels"]),
    ]

    # Compatibilité
    if not hasattr(cfg, "gpu_ids") or cfg.gpu_ids is None:
        cfg.gpu_ids = [0]
    if not hasattr(cfg, "seed") or cfg.seed is None:
        cfg.seed = SEED
    if not hasattr(cfg, "device") or cfg.device is None:
        cfg.device = "cuda" if torch.cuda.is_available() else "cpu"

    return cfg
```

Entraîner un fold (exemple pour fold 1)


```python
import gc
gc.collect()
torch.cuda.empty_cache()

fold = fold_info[0]  # fold 1
cfg_fold = prepare_cfg(cfg_base, fold["fold_dir"])

datasets = [build_dataset(cfg_fold.data.train)]
model = build_detector(cfg_fold.model)
set_random_seed(SEED)

train_detector(
    model,
    datasets,
    cfg_fold,
    distributed=False,
    validate=True
)
```

```python
#tester le fold (exemple pour fold 1)
from mmcv.parallel import MMDataParallel

test_dataset = build_dataset(cfg_fold.data.test, dict(test_mode=True))
test_loader = build_dataloader(
    test_dataset,
    samples_per_gpu=BATCH_SIZE,
    workers_per_gpu=0,
    dist=False,
    shuffle=False
)

best_ckpt = os.path.join(cfg_fold.work_dir, "epoch_50.pth")
if os.path.exists(best_ckpt):
    load_checkpoint(model, best_ckpt, map_location="cpu")
else:
    raise FileNotFoundError(f"{best_ckpt} not found!")

model = MMDataParallel(model, device_ids=[0])
model.eval()

outputs = single_gpu_test(model, test_loader)
results = test_dataset.evaluate(outputs, metric="bbox")

map50 = results["bbox_mAP_50"]
map50_95 = results["bbox_mAP"]
print(f"✅ Fold {fold['fold_idx']} | mAP50={map50:.4f} | mAP50-95={map50_95:.4f}")
```
